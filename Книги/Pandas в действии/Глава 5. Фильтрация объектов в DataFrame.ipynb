```python
import pandas as pd
```

```python
pd.read_csv("employees.csv")
```

```python
employees = pd.read_csv("employees.csv", parse_dates = ["Start Date"])

employees
```

```python
employees.info()
```

"""---5.1.1 Преобразование типов данных с помощью метода astype---"""

```python
employees["Mgmt"].astype(bool) # создали объект Series. значение NaN преобразовалось в True.
```

```python
employees["Mgmt"] = employees["Mgmt"].astype(bool) # это уже объект DataFrame
```

```python
employees.tail()
```

```python
employees.info()
```

```python
employees["Salary"].fillna(0).astype(int).tail()
```

```python
employees["Salary"] = employees["Salary"].fillna(0).astype(int)
```

```python
employees.nunique()
```

```python
employees["Gender"].astype("category")
```

```python
employees["Gender"] = employees["Gender"].astype("category")
```

```python
employees.info()
```

```python
employees["Team"] = employees["Team"].astype("category")
```

```python
employees.info()
```

"""---5.3 ФИЛЬТРАЦИЯ ПО НЕСКОЛЬКИМ УСЛОВИЯМ---"""

"""---5.3.1 Условие И---"""

```python
is_female = employees["Gender"] == "Female"
```

```python
in_biz_dev = employees["Team"] == "Business Dev"
```

```python
employees[is_female & in_biz_dev].head() # & - оператор "И"
```

```python
is_maneger = employees["Mgmt"]

employees[is_female & in_biz_dev & is_maneger].head()
```

"""---5.3.2 Условие ИЛИ---"""

```python
earning_below_40k = employees["Salary"] < 40000
started_after_2015 = employees["Start Date"] > "2015-01-01"
```

```python
employees[earning_below_40k | started_after_2015].tail() # | - оператор "ИЛИ"
```

"""---5.3.3 Логистическое отрицание---"""

```python
my_series = pd.Series([True, False, True])
my_series
```

```python
~my_series # меняет булев признак местами. False = True, True = False.
```

```python
employees[employees["Salary"] < 100000].head()
```

```python
employees[~(employees["Salary"] > 100000)].head() # все, что после "~" нужно помещать в (), чтобы не нарушить послежовательность выполнения операции.
```
