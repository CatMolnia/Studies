"""---5.3.3 Логистическое отрицание---"""

```python
my_series = pd.Series([True, False, True])
my_series
```

```python
~my_series # меняет булев признак местами. False = True, True = False.
```

```python
employees[employees["Salary"] < 100000].head()
```

```python
employees[~(employees["Salary"] > 100000)].head() # все, что после "~" нужно помещать в (), чтобы не нарушить послежовательность выполнения операции.
```

---5.4.4 Обработка пустых значений---

```python
employees = pd.read_csv(
    "employees.csv",
    parse_dates=["Start Date"]
)

employees
```

```python
employees.dropna() # удаляем строки в которых есть минимум одно пропущенное значение
```

```python
employees.dropna(how="all").tail() # how = all - удаление строк, в которых все значения = пропущенные
```

```python
employees.dropna(subset=["Gender"]).tail() # удаляем пустые значения из конкретного столбца
```

```python
employees.dropna(subset=["Start Date", "Salary"]).head() # удаляем пустые значения из нескольких указанных столбцов
```

```python
employees.dropna(thresh=4).head() # удаляем строку, только если в ней меньше чем 4ре заполненых значения. Если заполненных ≥ 4 — строка остаётся.
```

"""---5.5 РЕШЕНИЕ ПРОБЛЕМ ДУБЛИКАТОВ---"""

"""---5.5.1 Метод duplicated---"""

```python
employees["Team"].head()
```

```python
employees["Team"].duplicated().head() # отмечаем первое вхождение дубликата как False, следующие как True
```

```python
employees["Team"].duplicated(keep="first").head() # keep - указывает, какое из дублирующих значений оставлять. first - первое из дубликата = не дубликат
```

```python
employees["Team"].duplicated(keep="last") # keep - указывает, какое из дублирующих значений оставлять. last - последнее вхождение = не дубликат
```

```python
(~employees["Team"].duplicated()).head()
```

```python
first_one_in_team = ~employees["Team"].duplicated()

employees[first_one_in_team]
```

"""---5.5.2 Метод drop_duplicates---"""

```python
employees.drop_duplicates(subset=["Team"]) # альтернатива предыдущему шагу. В данном случае True = первое вхождение (дубликат удаляется)
```

```python
employees.drop_duplicates(subset=["Team"], keep="last") # в данном случае последние из дубликотов = True.
```

```python
employees.drop_duplicates(subset=["First Name"], keep=False) # оставляем только уникальные записи (не первое\последнее вхождение)
```

```python
name_is_douglas = employees["First Name"] == "Douglas"
is_male = employees["Gender"] == "Male"
employees[name_is_douglas & is_male]
```

```python
employees.drop_duplicates(subset=["Gender", "Team"]).head() # проверка на совпадение по нескольким столбцам. True = первое вхождение
```
