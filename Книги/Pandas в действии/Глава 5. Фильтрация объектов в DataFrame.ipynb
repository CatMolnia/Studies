```python
import pandas as pd
```

```python
pd.read_csv("employees.csv")
```

```python
employees = pd.read_csv("employees.csv", parse_dates = ["Start Date"])

employees
```

```python
employees.info()
```

"""---5.1.1 Преобразование типов данных с помощью метода astype---"""

```python
employees["Mgmt"].astype(bool) # создали объект Series. значение NaN преобразовалось в True.
```

```python
employees["Mgmt"] = employees["Mgmt"].astype(bool) # это уже объект DataFrame
```

```python
employees.tail()
```

```python
employees.info()
```

```python
employees["Salary"].fillna(0).astype(int).tail()
```

```python
employees["Salary"] = employees["Salary"].fillna(0).astype(int)
```

```python
employees.nunique()
```

```python
employees["Gender"].astype("category")
```

```python
employees["Gender"] = employees["Gender"].astype("category")
```

```python
employees.info()
```

```python
employees["Team"] = employees["Team"].astype("category")
```

```python
employees.info()
```
