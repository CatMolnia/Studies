"""---6.1 РЕГИСТР БУКВ И ПРОБЕЛЫ---"""

```python
import pandas as pd
```

```python
inspection = pd.read_csv("chicago_food_inspections.csv")

inspection
```

```python
inspection["Name"].head().values # получаем ndarray
```

```python
inspection['Name'].str # атрибут "srt" дает доступ к объекту StringMethods - для работы со строковыми значениями
```

```python
inspection["Name"].str.lstrip().head() # удаляем лишние пробелы слева
```

```python
inspection["Name"].str.rstrip().head() # удаляем лишние пробелы справа
```

```python
inspection["Name"].str.strip().head() # удаляем лишние пробелы И слева И справа сразу
```

```python
inspection["Name"] = inspection["Name"].str.strip() # создали новый объект Series без пробелов на основе столбца Name и перезаписали в существующий столбец
```

```python
inspection.columns
```

```python
# цикл по столбцам (переменная column) для удаления лишних пробелов разово везде
for column in inspection.columns:
  inspection[column] = inspection[column].str.strip() # inspection[column] = nspection["Name"]; inspection[column].str.strip() = inspection["Name"].str.strip()
```

```python
inspection["Name"].str.lower().head()
```

```python
inspection["Name"].str.capitalize().head() # приводим в верхний регистр только первую букву каждой строки столбца
```

```python
inspection["Name"].str.title().head() # приводим в верхний регистр первую букву каждого слова в строке столбца
```

"""---6.2 СРЕЗЫ СТРОКОВЫХ ЗНАЧЕНИЙ---"""

```python
inspection['Risk'].head()
```

```python
len(inspection)
```

```python
inspection["Risk"].unique()
```

```python
inspection = inspection.dropna(subset=["Risk"]) # удаляем пустые строки столбца "Risk"
```

```python
inspection["Risk"].unique()
```

```python
inspection = inspection.replace(
    to_replace="All",
    value="Risk 4 (Extreme)"
)
```

```python
inspection["Risk"].unique()
```

```python
inspection = inspection.replace(
    to_replace="Ris",
    value="Risk 1 (High +)"
)
```

```python
inspection["Risk"].unique()
```

"""---6.3 СРЕЗЫ СТРОКОВЫХ ЗНАЧЕНИЙ И ЗАМЕНА СИМВОЛОВ---"""

```python
inspection["Risk"].str.slice(5, 6).head() # для извлечения подстрок символов из строковых значений по позициям индексов можно воспользоваться методом slice. первый аргусмент - начало (включая), второй - конец (НЕ включая)
```

```python
inspection["Risk"].str[5:6].head() # аналог строки выше (синтаксис срезов списков языка Python)
```

```python
inspection["Risk"].str.slice(8).head() # извлекаем символы, начиная с позиции индекса 8 до конца строковых значений
```

```python
inspection["Risk"].str[8:].head() # аналог строки выше
```

```python
inspection["Risk"].str.slice(8, -1).head() # убираем последний знак строки из слайса
```

```python
inspection["Risk"].str[8:-1].head() # аналог строки выше
```

Для удаления закрывающих скобок можно также воспользоваться методом

Все методы str возвращают новый объект Series со своим собственным атрибутом str  .Это позволяет последовательно связывать цепочкой несколько
